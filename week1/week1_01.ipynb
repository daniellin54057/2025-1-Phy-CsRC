# 2. Then, extend to a more complex program

```python
import math

# ask the user to input three lengths of a triangle
a = float(input("Input the first length of the triangle: "))
b = float(input("Input the second length of the triangle: "))
c = float(input("Input the third length of the triangle: "))
print(f"The lengths of the triangle sides are: a = {a}, b = {b}, c = {c}")

# calculate the area of this triange
s = (a + b + c) / 2
area = math.sqrt(s * (s - a) * (s - b) * (s - c))

# output the area of this triangle 
print(f"The area of the triangle is: {area:.2f}")
```
