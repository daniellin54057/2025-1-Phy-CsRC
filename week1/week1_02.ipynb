## WSL Installation

Windows Subsystem for Linux (WSL) is a way to run a GNU/Linux environment
directly on Windows. 
You can see more details in the
[official documentation](https://docs.microsoft.com/en-us/windows/wsl/install).

To install WSL, you need to enable the Windows Subsystem
for Linux feature and Virtual Machine Platform in Windows Features.
You may have to reboot your system.



Today we will install Ubuntu in WSL. Open a terminal in Administrator mode and type these commands:

```shell
wsl --update
wsl --set-default-version 2
wsl --install -d Ubuntu
```

After the installation, you can start it by typing `wsl`
in the terminal. If it is the first time you run WSL, you will be asked to
input your username and password. **Remeber your password**.
You won't be able to gain administrator permissions in WSL once you lost it.
It's normal that you see nothing when you type your password.

## Miniconda Installation

[Miniconda](https://docs.conda.io/en/latest/miniconda.html) is a free and
open-source package management system and environment management system that
runs on multiple platforms.

To install Miniconda, you can follow the official installation guide and
paste these commands into your shell:


````python
mkdir -p ~/miniconda3
curl -fL https://repo.anaconda.com/miniconda/Miniconda3-latest-Linux-x86_64.sh > ~/miniconda3/miniconda.sh
bash ~/miniconda3/miniconda.sh -b -u -p ~/miniconda3
rm -rf ~/miniconda3/miniconda.sh
````

These commands will install Miniconda to `~/miniconda3`, so don't delete
this folder.

After installing, you should initialize Miniconda in your shell by:

```bash
~/miniconda3/bin/conda init bash
```

You may need to restart your shell to make it work.
Now, you can check it by running `conda` in the shell.

### Miniconda Tutorial

In the world of conda, we create an environment whenever we work on a
project. This is because we can install different packages and different
versions of packages in different environments, without
interfering with each other.

For the first few weeks, we don't need to create an environment
because we won't use external dependencies. You can just use the
`base` environment, which is used by default. However, this is not
a good idea once your homework evolves because dependencies will
be in a mass-up.

You can create an environment by:

```bash
conda create -n <env_name> <package1>=<version1> <package2>=<version2> ...
```

For example:

```bash
conda create -n lab1 python=3.11
```

After the creation, you can enter the environment by:

```bash
conda activate <env_name>
```

To exit a conda environment, run:

```bash
conda deactivate
```

To run Python code inside Jupyter notebook, you can install `ipykernel` in 
the environment:

```bash
conda install ipykernel
```

Then, choose the installed environment as the kernel in the button top right.
